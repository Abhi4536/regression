What is Simple Linear Regression?
It models the relationship between a dependent variable (Y) and one independent variable (X) using the equation Y = mX + c.

What are the key assumptions of Simple Linear Regression?

Linearity between X and Y

Independence of errors

Homoscedasticity (constant variance of errors)

Normal distribution of errors

What does the coefficient m represent in the equation Y = mX + c?
It is the slope, showing the change in Y for a one-unit change in X.

What does the intercept c represent in the equation Y = mX + c?
It is the value of Y when X = 0.

What is the purpose of the least squares method in Simple Linear Regression?
To minimize the sum of squared errors between predicted and actual Y values.

How is the coefficient of determination (R²) interpreted?
It shows the proportion of variance in Y explained by X.
R² = 1: Perfect fit; R² = 0: No explanatory power.

Multiple Linear Regression
What is Multiple Linear Regression?
A model that relates one dependent variable to two or more independent variables.

What is the main difference between Simple and Multiple Linear Regression?
Simple: One independent variable.
Multiple: Multiple independent variables.

What are the key assumptions of Multiple Linear Regression?

Linearity

Independence of errors

Homoscedasticity

Normal distribution of errors

No multicollinearity among predictors

What is heteroscedasticity, and how does it affect results?
When error variance is not constant — leads to inefficient estimates and unreliable significance tests.

How can you improve a model with high multicollinearity?

Drop correlated variables

Use Principal Component Analysis (PCA)

Apply Ridge or Lasso regression

What are common techniques for transforming categorical variables?

One-Hot Encoding

Label Encoding

Dummy Variables

What is the role of interaction terms in Multiple Linear Regression?
They model the combined effect of variables on the outcome.

How can the intercept's interpretation differ between Simple and Multiple Regression?

Simple: Y when X = 0

Multiple: Y when all predictors = 0 (which may not be meaningful)

What is the significance of the slope in regression analysis?
It shows how much the dependent variable changes for each unit change in a predictor.

How does the intercept provide context in regression models?
It gives a baseline value of the dependent variable when predictors are zero.

What are the limitations of using R² alone to evaluate a model?

Doesn’t account for overfitting

Doesn’t reflect predictor significance

Doesn’t improve with more meaningful variables unless adjusted R² is used

How would you interpret a large standard error for a coefficient?
Indicates high uncertainty and the predictor may not be statistically significant.

How can heteroscedasticity be identified and why is it important?

Use residual plots (fan or funnel shapes)

Must be addressed to ensure valid inference and error estimation

What if a model has high R² but low adjusted R²?
Likely includes irrelevant predictors, leading to overfitting.

Why is it important to scale variables in Multiple Regression?

Ensures equal weightage

Important for algorithms sensitive to scale (especially with regularization)

Polynomial Regression
What is Polynomial Regression?
A regression technique where the relationship is modeled as an nth-degree polynomial.

How does polynomial regression differ from linear regression?

Linear: Straight line fit

Polynomial: Curved (non-linear) fit

When is polynomial regression used?
When the data shows non-linear patterns that linear models can't capture.

Can polynomial regression be applied to multiple variables?
Yes, it becomes multivariate polynomial regression.

What are the limitations of polynomial regression?

Prone to overfitting

Difficult to interpret

Sensitive to outliers

How to evaluate model fit when choosing polynomial degree?

Cross-validation

Adjusted R²

AIC/BIC scores

Why is visualization important in polynomial regression?
Helps detect overfitting/underfitting and better understand the model's behavior.
